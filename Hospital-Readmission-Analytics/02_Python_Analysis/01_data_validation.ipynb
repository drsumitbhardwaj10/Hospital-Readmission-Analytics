## 1. Project Overview

This project analyzes hospital encounter data for patients with diabetes to understand patterns associated with hospital readmission.

The project follows an end-to-end healthcare analytics workflow:

**Raw Data → Data Validation → Data Cleaning → Exploratory Analysis → SQL Analysis → Power BI → Insights → Reporting**

---

## 2. Business Problem

Hospital readmissions can create additional healthcare utilization and may indicate opportunities to improve patient follow-up and care processes.

The primary analytical question for this project is:

> **What factors are associated with 30-day hospital readmission among patients with diabetes, and what insights can healthcare organizations derive from the data?**

---

## 3. Analytical Objectives

The project will investigate:

- The overall hospital readmission pattern
- 30-day readmission rates
- Readmission patterns by age and gender
- Relationship between length of stay and readmission
- Relationship between previous healthcare utilization and readmission
- Relationship between number of diagnoses and readmission
- Relationship between medication utilization and readmission
- Admission and discharge patterns
- Potential areas for further investigation

---

## 4. Dataset Description

The dataset contains hospital encounter records for patients with diabetes.

Each row represents a hospital encounter and contains demographic, clinical, utilization, and administrative variables.

The dataset contains information such as:

- Patient and encounter identifiers
- Demographic characteristics
- Admission information
- Length of hospital stay
- Laboratory procedures
- Medical procedures
- Medication information
- Previous healthcare utilization
- Diagnosis counts
- Readmission status

### Important Dataset Limitation

This is a historical dataset from hospitals in the United States covering the period 1999–2008.

Therefore, findings from this dataset should be interpreted as analytical findings from this dataset and **should not be assumed to represent current Indian hospitals or the Indian population**.
